```python
from keras import Sequential
from keras.layers import Dense, SimpleRNN
```

```python
model = Sequential()

model.add(SimpleRNN(3,input_shape=(4,5)))
model.add(Dense(1,activation='sigmoid'))

model.summary()
```

```python
print(model.get_weights()[0].shape)  # weights of the RNN layer
model.get_weights()[0]  # weights of the RNN layer
```

```python
print(model.get_weights()[1].shape)  # weights of the RNN layer
model.get_weights()[1]  # weights of the RNN layer
```

```python
print(model.get_weights()[2].shape)  # weights of the RNN layer
model.get_weights()[2]  # weights of the RNN layer
```

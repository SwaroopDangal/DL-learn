```python
import tensorflow
from tensorflow import keras
from keras import Sequential
from keras.layers import Dense
```

```python
model = Sequential()

model.add(Dense(2, input_dim=2, activation='sigmoid'))
model.add(Dense(1, activation='sigmoid'))
```

```python
model.summary()
```

```python
model.get_weights()
```

```python
new_weights = [
	np.array([[0.1, 0.1], [0.1, 0.1]], dtype=np.float32),  # dense1 kernel (2,2)
	np.array([0., 0.], dtype=np.float32),                 # dense1 bias (2,)
	np.array([[0.1], [0.1]], dtype=np.float32),            # dense2 kernel (2,1)
	np.array([0.0], dtype=np.float32)                      # dense2 bias (1,)
]
```

```python
model.set_weights(new_weights)
```
